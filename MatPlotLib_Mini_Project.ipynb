# **MINI PROJECT: VISUALIZING BUSINESS PERFORMANCE WITH MATPLOTLIB**


**BUSINESS SCENARIO:**

Imagine you have joined a small business as a junior data analyst. The business owner wants a simple Python-based report that provides an overview of revenue, customer growth, and product performance.
The business has provided six months of data. Your responsibility is to transform the numbers into a structured visual report and communicate your findings in language a business owner can understand.
This is an opportunity to combine the Matplotlib concepts you have learned and demonstrate your ability to apply them to a practical scenario.

**Project Dataset:**

months = [1, 2, 3, 4, 5, 6]

revenue = [150000, 165000, 158000, 190000, 215000, 240000]

customers = [80, 92, 88, 110, 125, 140]

product_a = [45, 50, 48, 60, 68, 75]

product_b = [35, 40, 42, 45, 52, 58]

**Understanding the variables**

months - Reporting months 1–6

revenue - Monthly business revenue in naira

customers - Number of customers recorded each month

product_a - Monthly sales quantity for Product A

product_b - Monthly sales quantity for Product B

For this project, assume that product values represent units sold. In a real project, you would verify whether these figures represent units sold, orders, or another measure.


```python
#Import the Library
import matplotlib.pyplot as plt

#Define the dataset
months = [1, 2, 3, 4, 5, 6]
revenue = [150000, 165000, 158000, 190000, 215000, 240000]
customers = [80, 92, 88, 110, 125, 140]
product_a = [45, 50, 48, 60, 68, 75]
product_b = [35, 40, 42, 45, 52, 58]

#To create a customized line chart for monthly revenue trend
plt.plot(months, revenue,
        color ="purple",
        marker = "o")
plt.title("Monthly Revenue Trend")  #adds chart title
plt.xlabel("Month")  #adds label to the x-axis
plt.ylabel("Revenue(₦)") #adds label to the y-axis
plt.xlim(1, 6)  #sets the x_axis range from month 1 - 6

# to display the full chart
plt.show()
```

```python
# to create line chart for customer growth
plt.plot(months, customers,
        color ="orange",
        linestyle = "--",
        marker = "o")
plt.title("Customer Growth")
plt.xlabel("Month")
plt.ylabel("Customers")
plt.xlim(1, 6)

plt.show()
```

```python
#to create line chart for product performance comparison
plt.plot(months,product_a,
        color = "blue",
         marker = "o",
        label = "Product A")
plt.plot(months, product_b,
        color = "green",
         marker = "o",
        label = "Product B")

plt.title("Product Performance Comparison")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.xlim(1, 6)
plt.legend() #adds legend to differentiate the products

plt.show()
```

```python
#to organize the charts into one figure using subplot()
#to create a figure with 3 charts using 2row - 2 column layout
fig, axes = plt.subplots(2, 2, figsize=(14,9))

#---------------------------------
# CHART 1: MONTHLY REVENUE TREND
# ---------------------------------

axes[0, 0].plot(months,revenue,
                color="purple",
                marker="o")

# To add the title and axis labels
axes[0, 0].set_title("Monthly Revenue Trend")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue (₦)")

# ---------------------------
# CHART 2: CUSTOMER GROWTH
# ---------------------------

axes[0, 1].plot(months, customers,
                color="blue",
                marker="s")

# to add the title and axis labels
axes[0, 1].set_title("Monthly Customer Growth")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Number of Customers")

# ----------------------------------------
# CHART 3: PRODUCT PERFORMANCE COMPARISON
# ----------------------------------------

axes[1, 0].plot(months,product_a,
                color="green",
                marker="o",
                label="Product A")

axes[1, 0].plot(months, product_b,
                color="orange",
                marker="s",
                label="Product B")

# To add the title, labels, and legend
axes[1, 0].set_title("Product Performance Comparison")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Units Sold")
axes[1, 0].legend()

# To add an overall title to the figure
fig.suptitle("Business Performance Visualization")

#To adjust spacing between the charts
fig.tight_layout()

#To display the complete figure
plt.show()
```

**Observations:**
1. Revenue rose from ₦150,000 in month 1 to ₦240,000 in month 6.
The overall increase is ₦90,000 while the percentage increase is 60%.
This shows a positive revenue trend across the reporting period, although revenue declined from ₦165,000 in month 2 to ₦158,000 in month 3.

2. The number of customers increased from 80 in month 1 to 140 in month 6. Overall increase is 60 customers while the percentage increase is 75%.
Customer numbers also declined slightly in month 3, falling from 92 to 88, before increasing in subsequent months.

3. Product A sold 75 units in month 6, compared with 58 units for Product B.
Product A therefore sold 17 more units than Product B in the final month. It also recorded higher unit sales in every month of the dataset.

4. Revenue and customer numbers both increased overall. Both also declined in month 3 compared with month 2.
This is an observed pattern, not proof that acquiring more customers caused revenue to increase. The data does not include purchase frequency, average transaction value, pricing, discounts, or customer spending.
